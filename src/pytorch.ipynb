```python
# What does dim = 0, 1, 2 mean for tnsr3D.mean(dim=1)?
# It specifies the dimension along which the operation (e.g., mean, exp) is applied.

# For example, to compute the mean across rows or columns in a tensor.

# Suppose: tnsr3D.shape = (A, B, C)
# Then:
#   dim = 0 → corresponds to A
#   dim = 1 → corresponds to B
#   dim = 2 → corresponds to C

# When we apply an operation with dim = k, that dimension is reduced (collapsed).

# Example:
# tnsr3D.mean(dim=0) → collapses dimension A
# Resulting shape: (B, C)

# tnsr3D.mean(dim=1) → collapses dimension B
# Resulting shape: (A, C)

# tnsr3D.mean(dim=2) → collapses dimension C
# Resulting shape: (A, B)
```
