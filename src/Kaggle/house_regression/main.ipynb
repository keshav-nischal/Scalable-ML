```python
def drop_not_required_columns(data):
    columns_to_drop = [
        "GarageYrBlt",
        "1stFlrSF",
        "TotRmsAbvGrd",
        "GarageArea"
    ]
    data = data.drop(columns=columns_to_drop)
    return data
```

```python
# Missing value handling strategy:
# 1. Columns where NA means the feature is absent (PoolQC, Alley, Fence, Garage*, Bsmt*, etc.)
#    → fill with "None" for categorical or 0 for numeric values.
# 2. Numeric features with moderate missing values (e.g., LotFrontage)
#    → fill using median (often grouped by Neighborhood) since median is robust to outliers.
# 3. Categorical features with very few missing values (e.g., Electrical)
#    → fill with mode to preserve the original category distribution.
# This approach minimizes bias while retaining as much training data as possible.

# Future improvements for missing value handling:
# - Add missing indicator features to capture information carried by missingness.
# - Use group-based imputation (e.g., median LotFrontage within Neighborhood).
# - Explore advanced methods such as KNN imputation or model-based imputation
#   to estimate missing values using relationships between features.
# For now, simple strategies (None/0, median, mode) provide a stable baseline
# and are commonly sufficient for initial regression models.


missing = x_train.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]
print(missing)
```

```python
# Apply missing value imputation for ALL columns
# This ensures the pipeline is robust for test data which may have NA values in different columns

# how is categorical data handled:
# Nominal data is a type of categorical data where values represent labels or names with no inherent order or ranking. eg: Colors: red, blue, green
# ordinal data
# One-Hot Encoding (OHE) eg: Color = {Red, Blue} → [1,0], [0,1]. Suitable for nominal variables with low-to-moderate cardinality.
# b. Label Encoding

# Linear / Neural models: prefer one-hot or embeddings.

# ====================================================================================
# CATEGORICAL DATA ENCODING
# ====================================================================================
# Strategy: 
# - Ordinal features (with inherent ranking): Label Encoding with proper order
# - Nominal features (no inherent order): One-Hot Encoding
# - Binary features: 0/1 encoding
import pandas as pd

def preprocess_housing_data(x_train, reference_columns=None):
    """
    Preprocess Ames Housing dataset with full coverage:
    - Missing value handling (robust)
    - Ordinal + binary + nominal encoding
    - Optional column alignment for test data
    
    Parameters:
    - x_train (pd.DataFrame): input dataset
    - reference_columns (list): columns from training set (for test alignment)
    
    Returns:
    - df (pd.DataFrame): processed dataframe
    - columns (list): final columns (for reuse on test set)
    """
    
    df = x_train.copy()

    # =========================================================
    # 1. HANDLE MISSING VALUES
    # =========================================================
    
    # NA means "absence"
    na_as_none = [
        "PoolQC","MiscFeature","Alley","Fence","FireplaceQu",
        "GarageType","GarageFinish","GarageQual","GarageCond",
        "BsmtQual","BsmtCond","BsmtExposure","BsmtFinType1","BsmtFinType2",
        "MasVnrType"
    ]
    
    for col in na_as_none:
        if col in df.columns:
            df[col] = df[col].fillna("None")
    
    # Numeric → median
    num_cols = df.select_dtypes(include=['int64','float64']).columns
    for col in num_cols:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].median())
    
    # Remaining categorical → mode
    cat_cols = df.select_dtypes(include=['object']).columns
    for col in cat_cols:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].mode()[0])

    # =========================================================
    # 2. ORDINAL ENCODING
    # =========================================================
    
    ordinal_mappings = {
        'ExterQual': {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'ExterCond': {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'BsmtQual': {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'BsmtCond': {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'HeatingQC': {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'KitchenQual': {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'FireplaceQu': {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'GarageQual': {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'GarageCond': {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
        'PoolQC': {'None':0,'Fa':1,'TA':2,'Gd':3,'Ex':4},
        'BsmtExposure': {'None':0,'No':0,'Mn':1,'Av':2,'Gd':3},
        'BsmtFinType1': {'None':0,'Unf':1,'LwQ':2,'Rec':3,'BLQ':4,'ALQ':5,'GLQ':6},
        'BsmtFinType2': {'None':0,'Unf':1,'LwQ':2,'Rec':3,'BLQ':4,'ALQ':5,'GLQ':6},
        'GarageFinish': {'None':0,'Unf':1,'RFn':2,'Fin':3},
        'PavedDrive': {'N':0,'P':1,'Y':2},
        'Fence': {'None':0,'MnWw':1,'MnPrv':2,'GdWo':3,'GdPrv':4},
        'LotShape': {'IR3':0,'IR2':1,'IR1':2,'Reg':3},
        'LandSlope': {'Sev':0,'Mod':1,'Gtl':2},
        'Utilities': {'ELO':0,'NoSeWa':1,'NoSewr':2,'AllPub':3},
        'Functional': {'Sal':0,'Sev':1,'Maj2':2,'Maj1':3,'Mod':4,'Min2':5,'Min1':6,'Typ':7},
    }
    
    for col, mapping in ordinal_mappings.items():
        if col in df.columns:
            df[col] = df[col].map(mapping)
            df[col] = df[col].fillna(df[col].median())

    # =========================================================
    # 3. BINARY ENCODING (AUTO-DETECT)
    # =========================================================
    
    for col in df.select_dtypes(include='object').columns:
        unique_vals = set(df[col].dropna().unique())
        if unique_vals == {'Y','N'}:
            df[col] = df[col].map({'N':0, 'Y':1})

    # =========================================================
    # 4. ONE-HOT ENCODING (ALL REMAINING CATEGORICAL)
    # =========================================================
    
    remaining_cat_cols = df.select_dtypes(include='object').columns
    df = pd.get_dummies(df, columns=remaining_cat_cols, drop_first=False, dtype=int)

    # =========================================================
    # 5. ALIGN COLUMNS (FOR TEST DATA)
    # =========================================================
    
    if reference_columns is not None:
        df = df.reindex(columns=reference_columns, fill_value=0)
        return df, reference_columns

    return df, df.columns.tolist()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer

# Create pipeline with the defined preprocessing functions
pipe = Pipeline([
    ("drop_columns", FunctionTransformer(drop_not_required_columns, validate=False)),
    ("preprocess_data", FunctionTransformer(lambda X: preprocess_housing_data(X)[0], validate=False))
])

# Fit and transform training data
x_train_processed = pipe.fit_transform(x_train)
print("Pipeline created and applied successfully!")
print(f"Original shape: {x_train.shape}")
print(f"Processed shape: {x_train_processed.shape}")
```

```python
test_data = pd.read_csv("./Data/train.csv")
y_test = test_data["SalePrice"]
x_test = test_data.drop(columns = "SalePrice")
```

```python
# training and testing
from sklearn.linear_model import LinearRegression

# Apply preprocessing pipeline to both train and test data
x_train_processed = pipe.fit_transform(x_train)
x_test_processed = pipe.transform(x_test)

# Train the model on processed data
model = LinearRegression().fit(x_train_processed, y_train)

# Make predictions on test data
y_pred = model.predict(x_test_processed)
print("Model trained successfully!")
print(f"Train shape: {x_train_processed.shape}")
print(f"Test shape: {x_test_processed.shape}")
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)

print(f"MSE: {mse}")
print(f"RMSE: {rmse}")
print(f"MAE: {mae}")
```
