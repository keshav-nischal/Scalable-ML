```python
# Missing value handling strategy:
# 1. Columns where NA means the feature is absent (PoolQC, Alley, Fence, Garage*, Bsmt*, etc.)
#    → fill with "None" for categorical or 0 for numeric values.
# 2. Numeric features with moderate missing values (e.g., LotFrontage)
#    → fill using median (often grouped by Neighborhood) since median is robust to outliers.
# 3. Categorical features with very few missing values (e.g., Electrical)
#    → fill with mode to preserve the original category distribution.
# This approach minimizes bias while retaining as much training data as possible.

# Future improvements for missing value handling:
# - Add missing indicator features to capture information carried by missingness.
# - Use group-based imputation (e.g., median LotFrontage within Neighborhood).
# - Explore advanced methods such as KNN imputation or model-based imputation
#   to estimate missing values using relationships between features.
# For now, simple strategies (None/0, median, mode) provide a stable baseline
# and are commonly sufficient for initial regression models.


missing = x_train.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]
print(missing)
```

```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error

class HousingPreprocessor(BaseEstimator, TransformerMixin):
    
    def __init__(self, skew_threshold=1):
        self.skew_threshold = skew_threshold

    def fit(self, X, y=None):
        processed, self.feature_names_ = self._process(X, is_fit=True)
        return self

    def transform(self, X, y=None):
        processed, _ = self._process(X, is_fit=False)

        missing_cols = {
            col: pd.Series(0, index=processed.index)
            for col in self.feature_names_
            if col not in processed.columns
        }
        if missing_cols:
            processed = pd.concat(
                [processed, pd.DataFrame(missing_cols, index=processed.index)],
                axis=1
            )

        processed = processed[self.feature_names_]
        return processed

    # ------------------------------------------------------------------
    # Internal helpers
    # ------------------------------------------------------------------

    def _drop_columns(self, df):
        columns_to_drop = ["GarageYrBlt", "1stFlrSF", "TotRmsAbvGrd", "GarageArea"]
        cols_present = [c for c in columns_to_drop if c in df.columns]
        return df.drop(columns=cols_present)

    def _apply_skew_transform(self, df, is_fit):
        """
        Only runs on continuous numeric columns — called BEFORE OHE
        so binary 0/1 dummies are never touched.
        """
        num_cols = df.select_dtypes(include='number').columns

        if is_fit:
            self.right_skew_cols_ = []
            self.left_skew_info_  = {}

            for col in num_cols:
                skew_val = df[col].skew()

                if skew_val > self.skew_threshold:
                    if (df[col] >= 0).all():
                        self.right_skew_cols_.append(col)
                        df[col] = np.log1p(df[col])

                elif skew_val < -self.skew_threshold:
                    max_val = df[col].max()
                    self.left_skew_info_[col] = max_val
                    df[col] = np.log1p(max_val - df[col])
        else:
            for col in self.right_skew_cols_:
                if col in df.columns:
                    df[col] = np.log1p(df[col])

            for col, max_val in self.left_skew_info_.items():
                if col in df.columns:
                    df[col] = np.log1p(max_val - df[col])

        return df

    def _process(self, X, is_fit):
        df = X.copy()

        # STEP 1: Drop unwanted columns
        df = self._drop_columns(df)

        # STEP 2: Missing value handling
        if 'MSSubClass' in df.columns:
            df['MSSubClass'] = df['MSSubClass'].astype(str)

        na_as_none = [
            "PoolQC", "MiscFeature", "Alley", "Fence", "FireplaceQu",
            "GarageType", "GarageFinish", "GarageQual", "GarageCond",
            "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1",
            "BsmtFinType2", "MasVnrType"
        ]
        for col in na_as_none:
            if col in df.columns:
                df[col] = df[col].fillna("None")

        if 'MasVnrArea' in df.columns:
            df['MasVnrArea'] = df['MasVnrArea'].fillna(0)

        if 'GarageYrBlt' in df.columns:
            df['GarageYrBlt'] = df['GarageYrBlt'].fillna(0)

        num_cols = df.select_dtypes(include=['int64', 'float64']).columns
        for col in num_cols:
            if df[col].isnull().any():
                df[col] = df[col].fillna(df[col].median())

        cat_cols = df.select_dtypes(include=['object']).columns
        for col in cat_cols:
            if df[col].isnull().any():
                df[col] = df[col].fillna(df[col].mode()[0])

        # STEP 3: Ordinal encoding
        ordinal_mappings = {
            'ExterQual':    {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'ExterCond':    {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'BsmtQual':     {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'BsmtCond':     {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'HeatingQC':    {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'KitchenQual':  {'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'FireplaceQu':  {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'GarageQual':   {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'GarageCond':   {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5},
            'PoolQC':       {'None':0,'Fa':1,'TA':2,'Gd':3,'Ex':4},
            'BsmtExposure': {'None':0,'No':0,'Mn':1,'Av':2,'Gd':3},
            'BsmtFinType1': {'None':0,'Unf':1,'LwQ':2,'Rec':3,'BLQ':4,'ALQ':5,'GLQ':6},
            'BsmtFinType2': {'None':0,'Unf':1,'LwQ':2,'Rec':3,'BLQ':4,'ALQ':5,'GLQ':6},
            'GarageFinish': {'None':0,'Unf':1,'RFn':2,'Fin':3},
            'PavedDrive':   {'N':0,'P':1,'Y':2},
            'Fence':        {'None':0,'MnWw':1,'MnPrv':2,'GdWo':3,'GdPrv':4},
            'LotShape':     {'IR3':0,'IR2':1,'IR1':2,'Reg':3},
            'LandSlope':    {'Sev':0,'Mod':1,'Gtl':2},
            'Utilities':    {'ELO':0,'NoSeWa':1,'NoSewr':2,'AllPub':3},
            'Functional':   {'Sal':0,'Sev':1,'Maj2':2,'Maj1':3,'Mod':4,'Min2':5,'Min1':6,'Typ':7},
        }
        for col, mapping in ordinal_mappings.items():
            if col in df.columns:
                df[col] = df[col].map(mapping)
                df[col] = df[col].fillna(df[col].median())

        # STEP 4: Binary encoding
        for col in df.select_dtypes(include='object').columns:
            if set(df[col].dropna().unique()) == {'Y', 'N'}:
                df[col] = df[col].map({'N': 0, 'Y': 1})

        # STEP 5: ✅ Skew correction BEFORE OHE
        # At this point df has only continuous numerics + remaining categoricals.
        # No binary 0/1 dummy columns exist yet, so skew detection is clean.
        df = self._apply_skew_transform(df, is_fit=is_fit)

        # STEP 6: One-hot encoding AFTER skew correction
        remaining_cat = df.select_dtypes(include='object').columns.tolist()
        df = pd.get_dummies(df, columns=remaining_cat, drop_first=False, dtype=int)

        return df, df.columns.tolist()
```

```python
# from sklearn.linear_model import LinearRegression, Ridge
# from sklearn.pipeline import Pipeline

# pipeline = Pipeline(steps=[
#     ('preprocessor', HousingPreprocessor(skew_threshold=1)),
#     ('model', Ridge(alpha=10))
# ])
# y_train_log = np.log1p(y_train)
# pipeline.fit(x_train, y_train_log)

# y_pred_log = pipeline.predict(x_test)
# y_pred = np.expm1(y_pred_log)  # reverse the log transform
from xgboost import XGBRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.pipeline import Pipeline
import numpy as np

pipeline = Pipeline(steps=[
    ('preprocessor', HousingPreprocessor(skew_threshold=1)),
    ('model', GradientBoostingRegressor(
        n_estimators=1000,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        random_state=42
    ))
])

y_train_log = np.log1p(y_train)

pipeline.fit(x_train, y_train_log)

y_pred_log = pipeline.predict(x_test)
y_pred = np.expm1(y_pred_log)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)

print(f"MSE: {mse}")
print(f"RMSE: {rmse}")
print(f"MAE: {mae}")
```
