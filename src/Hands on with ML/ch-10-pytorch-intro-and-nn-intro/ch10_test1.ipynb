```python
import torch
import time

# -----------------------
# Config
# -----------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
size = 5000   # increase to 10000+ for clearer GPU advantage
runs = 50

print(f"Using device: {device}")

# -----------------------
# Create tensors
# -----------------------
cpu_tensor = torch.randn(size, size)
gpu_tensor = cpu_tensor.to(device)

# -----------------------
# CPU Benchmark
# -----------------------
# warmup
for _ in range(10):
    cpu_tensor * cpu_tensor

start = time.time()
for _ in range(runs):
    cpu_tensor @ cpu_tensor
cpu_time = (time.time() - start) / runs

# -----------------------
# GPU Benchmark
# -----------------------
if device == "cuda":
    # warmup
    for _ in range(10):
        gpu_tensor * gpu_tensor

    torch.cuda.synchronize()

    start = time.time()
    for _ in range(runs):
        gpu_tensor @ gpu_tensor
    torch.cuda.synchronize()

    gpu_time = (time.time() - start) / runs
else:
    gpu_time = None

# -----------------------
# Results
# -----------------------
print(f"\nTensor size: {size} x {size}")
print(f"CPU time per op: {cpu_time:.6f} sec")

if gpu_time is not None:
    print(f"GPU time per op: {gpu_time:.6f} sec")
    print(f"Speedup (CPU/GPU): {cpu_time / gpu_time:.2f}x")
else:
    print("CUDA not available")
```

```python
#10
# search vs slope
# +/- the steps to get min cost function vs step big/small based on gradient (grad*lr)

# gradient 
x = torch.tensor(10.0, requires_grad=True)
lr = 0.01

for i in range(1000):
    f_x = x ** 2 + 10          # recompute from current x
    f_x.backward()        # compute gradient at current x

    with torch.no_grad():
        x -= lr * x.grad  # update x only

    x.grad.zero_()

print(f_x, x)
```

```python
#16, 14 gradient with detach(not no_grad())
x = torch.tensor(10.0, requires_grad=True)
lr = 0.01
x_detached = x.detach()
for i in range(1000):
    f_x = x ** 2 + 10          # recompute from current x
    f_x.backward()        # compute gradient at current x


    x_detached -= lr * x.grad  # update x only

    x.grad.zero_()

print(f_x, x)
# we can use clone to make seperate tensor, if we change x_detach, x will change
```
