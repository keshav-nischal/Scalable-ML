```python
# Lab 1.1 — See saturation for yourself. Plot the sigmoid function and its derivative on the same axes over a wide input range. Mark the regions where the derivative is effectively zero. Wonder: at what input magnitude does the gradient practically vanish? Repeat for tanh — why is a mean of 0 supposed to behave "slightly better" than a mean of 0.5?

#IMP: what is sigmoid function and what is its derivative, why is it useful, what charatersticks does it have which makes it useful
input = np.linspace(-10, 10, 500)          # symmetric, fine-grained

def sigmoid_fn(x):
    return 1 / (1 + np.exp(-x))

sigmoid = sigmoid_fn(input)

def sigmoid_derivative(x):
    sig = sigmoid_fn(x)
    return sig * (1 - sig)

sigmoid_der = sigmoid_derivative(input)

print(input)
print(sigmoid)
print(sigmoid_der)


plt.plot(input, sigmoid)
plt.plot(input, sigmoid_der)
```
